# Digit Recognition — Data Exploration & Results


```python
import sys
sys.path.insert(0, '..')

import torch
import matplotlib.pyplot as plt
import numpy as np
from torchvision import datasets, transforms

transform = transforms.Compose([transforms.ToTensor()])
mnist = datasets.MNIST('../data', train=True, download=True, transform=transform)
print(f'Training samples: {len(mnist):,}')
```

```python
fig, axes = plt.subplots(2, 5, figsize=(12, 5))
for digit in range(10):
    idx = next(i for i, (_, y) in enumerate(mnist) if y == digit)
    img, label = mnist[idx]
    ax = axes[digit // 5][digit % 5]
    ax.imshow(img.squeeze(), cmap='gray')
    ax.set_title(f'Digit: {label}', fontsize=12)
    ax.axis('off')
plt.suptitle('MNIST Sample Images', fontsize=14, fontweight='bold')
plt.tight_layout()
plt.savefig('../assets/sample_digits.png', dpi=150, bbox_inches='tight')
plt.show()
```

```python
import json
from pathlib import Path

summary_path = Path('../models/training_summary.json')
if summary_path.exists():
    with open(summary_path) as f:
        summary = json.load(f)
    print(f"Test accuracy : {summary['test_accuracy']:.4f} ({summary['test_accuracy']*100:.2f}%)")
    print(f"Best val acc  : {summary['best_val_accuracy']:.4f}")
    print()
    print('Per-class accuracy:')
    for digit, acc in summary['per_class_accuracy'].items():
        bar = '█' * int(acc * 40)
        print(f'  {digit}: {bar:<40} {acc*100:.2f}%')
else:
    print('Run training first: python scripts/run_training.py')
```

```python
import seaborn as sns

if summary_path.exists():
    cm = np.array(summary['confusion_matrix'])
    fig, ax = plt.subplots(figsize=(10, 8))
    sns.heatmap(cm, annot=True, fmt='d', cmap='Blues',
                xticklabels=range(10), yticklabels=range(10), ax=ax)
    ax.set_xlabel('Predicted', fontsize=12)
    ax.set_ylabel('True', fontsize=12)
    ax.set_title('Confusion Matrix — Test Set', fontsize=14)
    plt.tight_layout()
    plt.savefig('../assets/confusion_matrix.png', dpi=150, bbox_inches='tight')
    plt.show()
```
